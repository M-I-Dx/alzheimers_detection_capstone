## 01 - process - testset
In this notebook, we will process the original test set data, mapping file ids to text content 

```python
import pandas as pd
```

```python
# please point to correct downloaded transcribed folders of the original dataset
data_path = "../../../data/" 
path = data_path + "Test Dataset/Diagnosis Testing dataset/diagnosis/test-dist/"
```

```python
file1 = "test_results_task1.csv"
```

```python
print("Reading testset file...")
test_df = pd.read_csv(path + file1)
print("Completed")
test_df.fillna("", inplace=True)
test_df.head()
```

```python
print("Rows count =", len(test_df))
```

```python
# reading txt files
def read_txt(path, fname):
    """
        return the content of the txt file in string format
    """
    if not fname.endswith(".txt"):
        fname = fname + ".txt"
        
    try:
        with open(path + fname, encoding="utf-8") as f:
            lines = f.readlines()
            # print("Line len =", len(lines)) #, end="\r")
            text = "\n".join(lines)
            return text
    except Exception as e:
        print(f"Error reading file {path + fname}. Error = {e}")
 
```

```python
txt_path = data_path + "Transcribed Data-2024-Feb-13/Diagnosis Testing Data/"
read_txt(txt_path, "adrsdt15")
```

```python
# apply to dataframe to read txt content
print("Reading txt files....")
test_df["text"] = test_df["ID"].apply(lambda fname: read_txt(txt_path, fname))
print("Completed")
test_df.head()
```

```python
# write to csv file
output_path = data_path + "splits/"
test_df.to_csv(output_path + "test_text.csv",
               index=False
              )
print("Completed")
```

```python
# reformat task1 ground truth (0: non-AD, 1: AD)
task1_gt_fname = data_path + "Ground Truth for test sets/task1.csv"
task1_df = pd.read_csv(task1_gt_fname)
task1_df.head()
```

```python
fn = lambda label: 0 if label.lower().strip() == "control" else 1
assert fn("Control") == 0
assert fn("ProbableAD") == 1
print("Pass")
```

```python
# apply to dataframe
task1_df["target"] = task1_df["Dx"].apply(fn)
task1_df.head()
```
